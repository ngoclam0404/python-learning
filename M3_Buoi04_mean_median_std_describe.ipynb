```python
import pandas as pd
import numpy as np

url = ("https://raw.githubusercontent.com/jxchen/Kaggle"
       "/master/Give%20Me%20Some%20Credit/cs-training.csv")
df_raw = pd.read_csv(url, index_col=0)
df = df_raw.rename(columns={
    "SeriousDlqin2yrs":                    "vo_no",
    "RevolvingUtilizationOfUnsecuredLines": "ty_le_su_dung_tin_dung",
    "age":                                  "tuoi",
    "NumberOfTime30-59DaysPastDueNotWorse": "so_lan_tre_30_59_ngay",
    "DebtRatio":                            "ty_le_no",
    "MonthlyIncome":                        "thu_nhap_thang",
    "NumberOfOpenCreditLinesAndLoans":      "so_tai_khoan_vay",
    "NumberOfTimes90DaysLate":              "so_lan_tre_90_ngay",
    "NumberRealEstateLoansOrLines":         "so_tai_khoan_bat_dong_san",
    "NumberOfTime60-89DaysPastDueNotWorse": "so_lan_tre_60_89_ngay",
    "NumberOfDependents":                   "so_nguoi_phu_thuoc"
})
print(f"Shape: {df.shape} | NaN: {df.isnull().sum().sum()}")
```

```python
col = df["thu_nhap_thang"]

# Từng chỉ số riêng lẻ
print(f"Mean:   {col.mean():,.0f}")
print(f"Median: {col.median():,.0f}")
print(f"Mode:   {col.mode()[0]:,.0f}")

# ⚠️ BẪY QUAN TRỌNG — Pandas vs NumPy khác nhau:
print(f"\nPandas std (ddof=1): {col.std(ddof=1):,.2f}")
print(f"NumPy  std (ddof=0): {np.std(col):,.2f}")
print(f"Chênh nhau:          {col.std(ddof=1) - np.std(col):,.2f}")
# → Hai con số KHÁC NHAU dù cùng dữ liệu!

# describe() — tùy chỉnh percentile:
print("\n--- describe() tùy chỉnh ---")
print(col.describe(percentiles=[.1, .25, .5, .75, .9, .95, .99]))
```

```python
# So sánh std Pandas vs NumPy trên cột tuoi
col_tuoi = df["tuoi"]

pandas_std = col_tuoi.std(ddof=1)
numpy_std = np.std(col_tuoi)

print(f"Pandas std (ddof=1): {pandas_std:.4f}")
print(f"NumPy  std (ddof=0): {numpy_std:.4f}")
print(f"Chênh nhau:          {pandas_std - numpy_std:.6f}")

# value_counts cho vo_no
print("\n--- Tỷ lệ vỡ nợ ---")
print(df["vo_no"].value_counts(normalize=True).round(4))
```

```python
# Báo cáo thống kê 5 cột số
cols = ["tuoi", "thu_nhap_thang", "ty_le_su_dung_tin_dung",
        "ty_le_no", "so_lan_tre_90_ngay"]

for col in cols:
    s = df[col].dropna()
    mean = s.mean()
    median = s.median()
    std = s.std(ddof=1)
    print(f"\n{col}")
    print(f"  Mean: {mean:.2f} | Median: {median:.2f} | Std: {std:.2f}")
    print(f"  Min: {s.min():.2f} | Max: {s.max():.2f}")
    print(f"  P25: {s.quantile(0.25):.2f} | P75: {s.quantile(0.75):.2f}")
```

```python
# Phát hiện giá trị bất thường
print("--- tuoi = 0 ---")
print(f"Số hàng tuoi = 0: {(df['tuoi'] == 0).sum()}")

print("\n--- so_lan_tre_90_ngay = 96 hoặc 98 ---")
print(f"Số hàng = 96: {(df['so_lan_tre_90_ngay'] == 96).sum()}")
print(f"Số hàng = 98: {(df['so_lan_tre_90_ngay'] == 98).sum()}")

print("\n--- thu_nhap_thang = 0 ---")
print(f"Số hàng = 0: {(df['thu_nhap_thang'] == 0).sum()}")

print("\n--- value_counts so_lan_tre_90_ngay (top 10) ---")
print(df["so_lan_tre_90_ngay"].value_counts().head(10))
```
